```python
import huggingface_hub as hf
import os
import json

dataset_path = 'guychuk/open-prompt-injection'
# Pobierz dane z Hugging Face Hub
import pandas as pd

if not os.path.exists("../data/train.parquet"):
    os.makedirs("../data/")
    df = pd.read_parquet("hf://datasets/guychuk/open-prompt-injection/data/train-00000-of-00001.parquet")
    df.to_parquet("../data/train.parquet")
else:
    df = pd.read_parquet("../data/train.parquet")
df
```

```python
attack_types = df.groupby('attack_type').agg('first')
for attack_type, row in attack_types.iterrows():
    print(f"Attack type: {attack_type}")
    print(f"safe prompt: {row['normal_input']}")
    print(f"malicious prompt: {row['attack_input']}")
```

```python
attack_types = df.groupby('injected_task').agg('first')
for injected_task, row in attack_types.iterrows():
    print(f"Attack type: {injected_task}")
    print(f"safe prompt: {row['normal_input']}")
    print(f"malicious prompt: {row['attack_input']}")
```
